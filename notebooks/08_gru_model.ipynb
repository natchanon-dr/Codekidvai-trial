```python
print('== LSTM / GRU Comparison Contract Audit ==')
print()

# Compute GRU parameter count
import torch.nn as nn
def count_params(model):
    return sum(p.numel() for p in model.parameters())

def set_seeds(seed):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed)
    torch.use_deterministic_algorithms(True, warn_only=True)
    os.environ['PYTHONHASHSEED'] = str(seed)

class GRUSeqOnly(nn.Module):
    def __init__(self, n_features, gru_units, dropout_rate):
        super().__init__()
        self.gru     = nn.GRU(n_features, gru_units, batch_first=True)
        self.dropout = nn.Dropout(dropout_rate)
        self.fc      = nn.Linear(gru_units, 1)
    def forward(self, x, lengths):
        packed = nn.utils.rnn.pack_padded_sequence(
            x, lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, h_n = self.gru(packed)   # h_n: (1, B, H) -- no cell state (GRU differs from LSTM)
        h = h_n.squeeze(0)
        h = self.dropout(h)
        return torch.sigmoid(self.fc(h)).squeeze(1)

class GRUSeqTag(nn.Module):
    def __init__(self, n_features, n_tag, gru_units, dropout_rate):
        super().__init__()
        self.gru     = nn.GRU(n_features, gru_units, batch_first=True)
        self.dropout = nn.Dropout(dropout_rate)
        self.fc      = nn.Linear(gru_units + n_tag, 1)
    def forward(self, x, lengths, tag):
        packed = nn.utils.rnn.pack_padded_sequence(
            x, lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, h_n = self.gru(packed)
        h = h_n.squeeze(0)
        h = self.dropout(h)
        combined = torch.cat([h, tag], dim=1)
        return torch.sigmoid(self.fc(combined)).squeeze(1)

set_seeds(42)
gru_demo  = GRUSeqOnly(N_FEATURES, GRU_UNITS, DROPOUT_RATE)
gru_params = count_params(gru_demo)

# Reconstruct LSTM param count for comparison
class LSTMSeqOnly(nn.Module):
    def __init__(self, n_features, lstm_units, dropout_rate):
        super().__init__()
        self.lstm    = nn.LSTM(n_features, lstm_units, batch_first=True)
        self.dropout = nn.Dropout(dropout_rate)
        self.fc      = nn.Linear(lstm_units, 1)
    def forward(self, x, lengths):
        packed = nn.utils.rnn.pack_padded_sequence(
            x, lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, (h_n, _) = self.lstm(packed)
        h = h_n.squeeze(0)
        h = self.dropout(h)
        return torch.sigmoid(self.fc(h)).squeeze(1)

lstm_demo   = LSTMSeqOnly(N_FEATURES, GRU_UNITS, DROPOUT_RATE)
lstm_params = count_params(lstm_demo)

m4_ds = m4_manifest.get('dataset_stats', {})

contract_rows = [
    ('input_tensor_shape_train',  str(tuple(X_train_full.shape)),
                                  str(tuple(m4_ds.get('tensor_shape_train', [])))),
    ('input_tensor_shape_test',   str(tuple(X_test.shape)),
                                  str(tuple(m4_ds.get('tensor_shape_test', [])))),
    ('recurrent_units',           str(GRU_UNITS),   str(m4_config.get('lstm_units', '?'))),
    ('dropout_rate',              str(DROPOUT_RATE), str(m4_config.get('dropout_rate', '?'))),
    ('optimizer',                 'Adam',             m4_config.get('optimizer', 'Adam')),
    ('learning_rate',             str(LEARNING_RATE), str(m4_config.get('learning_rate', '?'))),
    ('batch_size',                str(BATCH_SIZE),    str(m4_config.get('batch_size', '?'))),
    ('max_epochs',                str(MAX_EPOCHS),    str(m4_config.get('max_epochs', '?'))),
    ('early_stop_patience',       str(EARLY_STOP_PATIENCE), str(m4_config.get('early_stop_patience', '?'))),
    ('seeds',                     str(SEEDS),         str(m4_config.get('seeds', '?'))),
    ('val_split_size',            str(VAL_SPLIT_SIZE), str(m4_config.get('val_split_size', '?'))),
    ('train_learners',            str(len(train_learners)), str(m4_ds.get('train_learners', '?'))),
    ('val_learners',              '2 (from train pool)', '2 (from train pool)'),
    ('test_learners',             str(len(test_learners)), str(m4_ds.get('test_learners', '?'))),
    ('trainable_params',          str(gru_params),    str(lstm_params)),
    ('recurrent_cell',            'GRU',              'LSTM'),
]

contract_df = pd.DataFrame(contract_rows, columns=['parameter','GRU (M5)','LSTM (M4)'])
contract_df['match'] = contract_df.apply(
    lambda r: 'DIFF' if r['GRU (M5)'] != r['LSTM (M4)'] else 'SAME', axis=1)

print(contract_df.to_string(index=False))
n_diffs = (contract_df['match'] == 'DIFF').sum()
print(f'\nDifferences from M4: {n_diffs}')
diffs = contract_df[contract_df['match'] == 'DIFF']
for _, row in diffs.iterrows():
    print(f'  {row["parameter"]}: GRU={row["GRU (M5)"]!r}  LSTM={row["LSTM (M4)"]!r}')
print()
print('NOTE: trainable_params and recurrent_cell differences are expected (cell architecture).')
print('All other parameters must match for fair comparison.')
```

```python
pre_checks = []
def pre_chk(name, passed, detail=''):
    pre_checks.append({'check':name,'result':'PASS' if passed else 'FAIL','detail':str(detail)})
    print(f'  {"OK" if passed else "FAIL"} {name}' + (f' -- {detail}' if detail else ''))

print('-- Pre-training checks --')

# 1: M2 input checksums in manifest
pre_chk('M2 input checksums in manifest',
    all(k in manifest.get('input_files',{}) for k in ['sequence_sha','attempt_sha','outcome_sha']),
    f"{len(manifest.get('input_files',{}))} entries")

# 2: Frozen split matches M4
m4_ds = m4_manifest.get('dataset_stats', {})
split_matches_m4 = (
    X_train_full.shape[0] == m4_ds.get('train_sequences', -1) and
    X_test.shape[0]       == m4_ds.get('test_sequences', -1)  and
    len(train_learners)   == m4_ds.get('train_learners', -1)  and
    len(test_learners)    == m4_ds.get('test_learners', -1)
)
pre_chk('Frozen split matches M4',
    split_matches_m4,
    f'train_seqs={X_train_full.shape[0]} (M4:{m4_ds.get("train_sequences","?")}), '
    f'test_seqs={X_test.shape[0]} (M4:{m4_ds.get("test_sequences","?")})')

# 3: Tensor shape matches M4
m4_train_shape = tuple(m4_ds.get('tensor_shape_train', []))
m4_test_shape  = tuple(m4_ds.get('tensor_shape_test', []))
pre_chk('Tensor shape matches M4',
    tuple(X_train_full.shape) == m4_train_shape and tuple(X_test.shape) == m4_test_shape,
    f'train={X_train_full.shape} (M4:{m4_train_shape})')

# 4: Feature list/order matches M4
m4_feature_names = m4_config.get('n_features', None)
gru_n_features   = N_FEATURES
pre_chk('Feature dimension matches M4',
    m4_feature_names is None or int(m4_feature_names) == gru_n_features,
    f'N_FEATURES={gru_n_features} (M4:{m4_feature_names})')

# 5: No learner overlap
overlap = set(train_learners) & set(test_learners)
pre_chk('No learner overlap', len(overlap)==0, f'overlap={len(overlap)}')

# 6: No NaN/Inf
no_nan = not(np.isnan(X_train_full).any() or np.isinf(X_train_full).any()
             or np.isnan(X_test).any() or np.isinf(X_test).any())
pre_chk('No NaN/Inf in tensors', no_nan)

# 7: Padding mask valid
pre_chk('Padding mask shape valid',
    mask_train_full.shape==X_train_full.shape[:2] and mask_test.shape==X_test.shape[:2],
    f'train_mask={mask_train_full.shape}  test_mask={mask_test.shape}')

# 8: No blacklisted outcome fields in X
m2_feats = manifest.get('parameters',{}).get('feature_names',[])
bl_in_x  = sorted(set(m2_feats) & BLACKLISTED_X_FIELDS)
pre_chk('No blacklisted outcome fields in X', len(bl_in_x)==0,
    f'blacklisted: {bl_in_x}' if bl_in_x else 'clean')

n_pre_fail = sum(1 for c in pre_checks if c['result']=='FAIL')
print(f'\n{len(pre_checks)-n_pre_fail}/{len(pre_checks)} pre-training checks passed')
if n_pre_fail > 0:
    raise RuntimeError(f'Pre-training validation FAILED ({n_pre_fail} checks).')
```

```python
def make_val_split(X, y, groups, seed, val_size):
    gss = GroupShuffleSplit(n_splits=1, test_size=val_size, random_state=seed)
    tr_idx, vl_idx = next(gss.split(X, y, groups))
    assert not (set(groups[tr_idx]) & set(groups[vl_idx])), 'SPLIT LEAKAGE'
    return tr_idx, vl_idx


def compute_metrics(y_true, y_prob, threshold=DECISION_THRESHOLD):
    y_pred = (y_prob >= threshold).astype(int)
    acc  = float(accuracy_score(y_true, y_pred))
    ncls = len(np.unique(y_true))
    if ncls < 2:
        prec=rec=f1=auc=None
    else:
        prec = float(precision_score(y_true, y_pred, zero_division=0))
        rec  = float(recall_score(y_true, y_pred, zero_division=0))
        f1   = float(f1_score(y_true, y_pred, zero_division=0))
        try:
            auc = float(roc_auc_score(y_true, y_prob))
        except Exception:
            auc = None
    cm = confusion_matrix(y_true, y_pred).tolist()
    return {'accuracy':acc,'precision':prec,'recall':rec,'f1':f1,
            'roc_auc':auc,'confusion_matrix':cm,'n_classes_in_test':int(ncls)}


def run_epoch(model, X_t, y_t, mask_t, optimizer, criterion, cw_tensor,
              batch_size, tag_t=None, training=True):
    model.train(training)
    n = X_t.shape[0]
    indices = list(range(n))
    if training: random.shuffle(indices)
    total_loss = 0.0; n_batches = 0
    with torch.set_grad_enabled(training):
        for start in range(0, n, batch_size):
            idx  = indices[start:start+batch_size]
            xb   = X_t[idx]; yb = y_t[idx].float(); mb = mask_t[idx]
            lens = torch.tensor(mb.sum(dim=1).clamp(min=1), dtype=torch.int64)
            preds = (model(xb, lens, tag_t[idx]) if tag_t is not None else model(xb, lens))
            weights = torch.where(yb==1, cw_tensor[1], cw_tensor[0])
            loss = (criterion(preds, yb) * weights).mean()
            if training:
                optimizer.zero_grad(); loss.backward(); optimizer.step()
            total_loss += loss.item(); n_batches += 1
    return total_loss / max(n_batches, 1)


def run_seed(seed, experiment, X_pool, y_pool, mask_pool, groups,
             X_te, y_te, mask_te, tag_pool=None, tag_te=None):
    set_seeds(seed)
    tr_idx, vl_idx = make_val_split(X_pool, y_pool, groups, seed, VAL_SPLIT_SIZE)
    X_tr = torch.tensor(X_pool[tr_idx]); y_tr = torch.tensor(y_pool[tr_idx])
    X_vl = torch.tensor(X_pool[vl_idx]); y_vl = torch.tensor(y_pool[vl_idx])
    mask_tr = torch.tensor(mask_pool[tr_idx]); mask_vl = torch.tensor(mask_pool[vl_idx])
    X_te_t  = torch.tensor(X_te); mask_te_t = torch.tensor(mask_te)

    classes = np.unique(y_pool[tr_idx])
    if len(classes) == 2:
        cw_vals = compute_class_weight('balanced', classes=classes, y=y_pool[tr_idx])
        cw = torch.tensor([cw_vals[0], cw_vals[1]], dtype=torch.float32)
    else:
        cw = torch.ones(2, dtype=torch.float32)

    tag_tr_t = tag_vl_t = tag_te_t = None
    if experiment == 'seq_tag' and tag_pool is not None:
        tag_tr_t = torch.tensor(tag_pool[tr_idx])
        tag_vl_t = torch.tensor(tag_pool[vl_idx])
        tag_te_t = torch.tensor(tag_te)
        model = GRUSeqTag(N_FEATURES, n_tag, GRU_UNITS, DROPOUT_RATE).to(DEVICE)
    else:
        model = GRUSeqOnly(N_FEATURES, GRU_UNITS, DROPOUT_RATE).to(DEVICE)

    optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
    criterion = nn.BCELoss(reduction='none')

    history = {'loss':[],'val_loss':[],'accuracy':[],'val_accuracy':[]}
    best_val_loss = float('inf'); patience_count = 0; best_state = None

    t0 = time.perf_counter()
    for epoch in range(MAX_EPOCHS):
        tr_loss = run_epoch(model, X_tr, y_tr, mask_tr, optimizer, criterion, cw,
                            BATCH_SIZE, tag_tr_t, training=True)
        vl_loss = run_epoch(model, X_vl, y_vl, mask_vl, optimizer, criterion, cw,
                            BATCH_SIZE, tag_vl_t, training=False)
        with torch.no_grad():
            lens_tr = mask_tr.sum(dim=1).clamp(min=1)
            lens_vl = mask_vl.sum(dim=1).clamp(min=1)
            p_tr = (model(X_tr, lens_tr, tag_tr_t) if tag_tr_t is not None
                    else model(X_tr, lens_tr)).numpy()
            p_vl = (model(X_vl, lens_vl, tag_vl_t) if tag_vl_t is not None
                    else model(X_vl, lens_vl)).numpy()
        tr_acc = float(accuracy_score(y_tr.numpy(), (p_tr>=DECISION_THRESHOLD).astype(int)))
        vl_acc = float(accuracy_score(y_vl.numpy(), (p_vl>=DECISION_THRESHOLD).astype(int)))
        history['loss'].append(tr_loss); history['val_loss'].append(vl_loss)
        history['accuracy'].append(tr_acc); history['val_accuracy'].append(vl_acc)
        # Early stopping monitors val_loss only -- test is NEVER used
        if vl_loss < best_val_loss:
            best_val_loss = vl_loss; patience_count = 0
            best_state = {k: v.clone() for k, v in model.state_dict().items()}
        else:
            patience_count += 1
            if patience_count >= EARLY_STOP_PATIENCE:
                break
    train_t = time.perf_counter() - t0
    if best_state:
        model.load_state_dict(best_state)

    model.eval()
    t1 = time.perf_counter()
    with torch.no_grad():
        lens_te = mask_te_t.sum(dim=1).clamp(min=1)
        y_prob = (model(X_te_t, lens_te, tag_te_t) if tag_te_t is not None
                  else model(X_te_t, lens_te)).numpy()
    inf_t = (time.perf_counter() - t1) / max(len(y_te), 1)

    m = compute_metrics(y_te, y_prob)
    n_params_model = count_params(model)
    m.update({'train_time_sec': round(train_t,4),
              'inf_time_per_seq_sec': round(inf_t,6),
              'epochs_trained': len(history['loss']),
              'trainable_params': n_params_model,
              'seed': seed, 'experiment': experiment,
              'n_sub_train_seqs': int(len(tr_idx)),
              'n_val_seqs': int(len(vl_idx)),
              'n_test_seqs': int(len(y_te)),
              'class_weight': cw.tolist(),
              'label_validity': 'pilot_only', 'label_source': 'proxy_behavioral',
              'roc_auc_note': ('computed' if m['n_classes_in_test']>=2
                               else 'ROC-AUC undefined -- test set contains only one class')})
    return {'seed':seed,'experiment':experiment,'metrics':m,'history':history,
            'y_pred_prob':y_prob.tolist(),'y_true':y_te.tolist(),'model':model}


print('EXP-A: GRU Sequence Only')
print(f'Running {len(SEEDS)} seeds: {SEEDS}')
print(f'GRU trainable params: {gru_params}  LSTM trainable params: {lstm_params}\n')

exp_a_results = []
for seed in SEEDS:
    print(f'  Seed {seed:3d} ...', end=' ', flush=True)
    r = run_seed(seed, 'seq_only',
                 X_train_full, y_train_full, mask_train_full, train_learner_groups,
                 X_test, y_test, mask_test)
    m = r['metrics']
    auc = f"{m['roc_auc']:.4f}" if m['roc_auc'] is not None else 'NA'
    f1v = f"{m['f1']:.4f}"      if m['f1']      is not None else 'NA'
    print(f"acc={m['accuracy']:.4f}  f1={f1v}  auc={auc}  epochs={m['epochs_trained']}  "
          f"train={m['train_time_sec']:.1f}s")
    exp_a_results.append(r)
print('\nEXP-A complete.')
```

```python
def summarise_seeds(results, label):
    keys = ['accuracy','precision','recall','f1','roc_auc',
            'train_time_sec','inf_time_per_seq_sec','epochs_trained','trainable_params']
    s = {'experiment':label, 'seeds':[r['seed'] for r in results]}
    for k in keys:
        vals = [r['metrics'][k] for r in results if r['metrics'].get(k) is not None]
        s[f'{k}_mean'] = round(float(np.mean(vals)),6) if vals else None
        s[f'{k}_std']  = round(float(np.std(vals)), 6) if vals else None
    s['per_seed'] = [r['metrics'] for r in results]
    return s

exp_a_summary = summarise_seeds(exp_a_results, 'EXP-A: GRU Sequence Only')
print('EXP-A Summary')
print(f"  accuracy       : {exp_a_summary['accuracy_mean']:.4f} +/- {exp_a_summary['accuracy_std']:.4f}")
print(f"  f1             : {exp_a_summary['f1_mean']} +/- {exp_a_summary['f1_std']}")
print(f"  roc_auc        : {exp_a_summary['roc_auc_mean']} +/- {exp_a_summary['roc_auc_std']}")
print(f"  epochs         : {exp_a_summary['epochs_trained_mean']:.1f} +/- {exp_a_summary['epochs_trained_std']:.1f}")
print(f"  trainable_params: {exp_a_summary['trainable_params_mean']}")
print('WARNING: pilot_only / proxy_behavioral -- pipeline validation only')
```

```python
all_checks = list(pre_checks)
def chk(name, passed, detail=''):
    all_checks.append({'check':name,'result':'PASS' if passed else 'FAIL','detail':str(detail)})

# 9: Test data excluded from early stopping
chk('Test data excluded from early stopping', True,
    'EarlyStopping monitors val_loss from val learners only')

# 10: Predictions within [0,1]
all_preds = [p for r in exp_a_results for p in r['y_pred_prob']]
chk('Predictions within [0,1]',
    all(0.0<=p<=1.0 for p in all_preds),
    f'min={min(all_preds):.4f}  max={max(all_preds):.4f}')

# 11: Prediction count matches test sequences
n_test = X_test.shape[0]
chk('Prediction count == test sequences',
    all(len(r['y_pred_prob'])==n_test for r in exp_a_results),
    f'expected={n_test}')

# 12: Timing non-negative
chk('Timing values non-negative',
    all(r['metrics']['train_time_sec']>=0 and r['metrics']['inf_time_per_seq_sec']>=0
        for r in exp_a_results))

# 13: All outputs labeled pilot_only
chk('All metrics labeled pilot_only',
    all(r['metrics'].get('label_validity')=='pilot_only' for r in exp_a_results))

# 14: Reproducibility
chk('Same-seed execution is reproducible', repro_ok, f'max_pred_diff={max_diff:.2e}')

# 15: LSTM/GRU comparison contract documented
contract_ok = 'contract_df' in dir() and len(contract_df) > 0
expected_diffs = {'recurrent_cell', 'trainable_params'}
actual_diffs   = set(contract_df[contract_df['match']=='DIFF']['parameter'].tolist()) if contract_ok else set()
unexpected_diffs = actual_diffs - expected_diffs
chk('LSTM/GRU comparison contract documented',
    contract_ok and len(unexpected_diffs)==0,
    f'documented {len(contract_df)} params, {len(actual_diffs)} diffs '
    f'(expected: {sorted(expected_diffs)}; unexpected: {sorted(unexpected_diffs)})')

result_df   = pd.DataFrame(all_checks)
n_hard_fail = sum(1 for c in all_checks if c['result']=='FAIL')

print('\n-- M5 Validation Summary --')
print(result_df.to_string(index=False))
print(f'\n{len(all_checks)-n_hard_fail}/{len(all_checks)} checks passed')

if n_hard_fail > 0:
    raise RuntimeError(f'M5 validation FAILED -- {n_hard_fail} check(s).')

print('\nM5 COMPLETE -- Pilot GRU pipeline validated.')
print('WARNING PILOT LIMITATIONS:')
print(f'  label_source=proxy_behavioral / label_validity=pilot_only')
print(f'  Learners={len(split_ledger)} total ({len(train_learners)} train, {len(test_learners)} test)')
print(f'  Tensors: train={X_train_full.shape}  test={X_test.shape}')
print(f'  GRU trainable params: {gru_params}  LSTM trainable params: {lstm_params}')
print(f'  These results are NOT final Chapter 4 conclusions.')
print(f'  Do NOT claim GRU superiority based on 10-learner pilot.')
```
